```python
class BankAccount:
      def __init__(self,name,account_no,balance):
            self.name = name
            self.account_no = account_no
            self.balance = balance
            self.Save()
            
      def Save(self):
        check="yes"
        import json
        with open("Data.json","r")as f:
                  for i in f:
                        a=json.loads(i)
                        try:
                              if a["name"]==self.name and a["account_no"]==self.account_no:
                                    raise Exception("this account already existed")
                        except Exception as e:
                              print(e)
                              check="no"
        if check=="yes":
          import json
          with open("Data.json","a")as f:
              json.dump({"name":self.name,"account_no":self.account_no,"balance":self.balance},f)
              f.write("\n")
              
      
      def Deposit(self,name,account_no,amount):
            try:
                  if type(amount)!=int:
                        raise Exception("please put amount in digit")
                  elif amount<0:
                        raise Exception("please put ammount in +ve")
                  elif amount==0:
                        raise Exception("please put some ammount")
            except Exception as e:
                  print(e)
            else:
                  li=[]
                  check="no"
                  import json
                  with open("Data.json","r")as f:
                        for i in f:
                              a=json.loads(i)
                              if a["name"]==name and a["account_no"]==account_no:
                                    a["balance"]+=amount
                                    check="yes"
                              li.append(a)
                                    
                              
                        if check=="no":
                              print("your account not found")
                  import json 
                  with open("Data.json","w")as f:
                        for i in li:
                              json.dump(i,f)
                              f.write("\n")      
                                    
                              
      def withdraw(self,name,account_no,amount):
            try:
                  if type(amount)!=int:
                        raise Exception("please put ammount in digit")
                  elif amount<0:
                        raise Exception("please put amount in +ve")
                  elif amount==0:
                        raise Exception("please put some ammount")
            except Exception as e:
                  print(e)
            else:
                  lis=[]
                  check="no"
                  import json
                  with open("Data.json","r")as f:
                        for i in f:
                              a=json.loads(i)
                              if a["name"]==name and a["account_no"]==account_no:
                                    a["balance"]-=amount
                                    check="yes"
                              lis.append(a)
                        if check=="no":
                              print("your account not found")
                  import json
                  with open("Data.json","w")as f:
                        for i in lis:
                              json.dump(i,f)
                              f.write("\n")    
                              
      def Display(self,name,account_no):
            import json
            with open("Data.json","r")as f:
                  check="no"
                  for i in f:
                        a=json.loads(i)
                        if a["name"]==name and a["account_no"]==account_no:
                              check="yes"
                              print(f"Name:{a["name"]}",f"Account_no:{a["account_no"]}",f"Balance:{a["balance"]}",sep="\n") 
                  if check=="no":
                        print("your account not found")             
                                    
                              
                                    
                  
            
                                    
                                          
```
